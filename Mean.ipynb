<b>Different ways to calculate mean (average) in Python.</b>

```python
import statistics

X = [1, 2, 3, 4]
avg_ex = statistics.mean(X)
print(avg_ex)
```

```python
from statistics import mean

X = [1, 2, 3, 4]
avg_ex = mean(X)
print(avg_ex)
```

```python
X = [1, 2, 3, 4]
avg_ex = sum(X) / len(X)
print(avg_ex)
```

The NumPy library is a powerful tool for numerical operations in Python. If you're working with large datasets or complex mathematical calculations, NumPy is often the best choice. It's much faster than standard Python lists for these operations.

```python
import numpy as np

X = [1, 2, 3, 4]
avg_ex = np.mean(X)
print(avg_ex)
```

SciPy is another scientific computing library that builds upon NumPy. While it's generally used for more advanced scientific and technical computing tasks, it also includes a function to calculate the mean.

```python
# install SciPy
!pip install scipy
```

```python
from scipy import stats

X = [1, 2, 3, 4]
avg_ex = stats.tmean(X)
print(avg_ex)  # 2.5
```

You can also use the pandas library, which is a leading library for data analysis.

```python
import pandas as pd

X = [1, 2, 3, 4]
series = pd.Series(X)
avg_ex = series.mean()
print(avg_ex)
```
